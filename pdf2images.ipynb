```python
import pymupdf
from pathlib import Path

year = "2025"
quarter = "q1"

pic_path = f"pic/{year}/{quarter}/"

sizes = ["small", "large", "superlarge", "medium"]

Path(pic_path).mkdir(parents=True, exist_ok=True)
```

```python
for size in sizes:
    pdf_path = f"data/{year}/{year}_{quarter}_{size}.pdf"
    print(f"{pdf_path}")
    
    doc = pymupdf.open(pdf_path)
    for page in doc:
        texts = page.get_text()
        images = page.get_images()
        print(f"Page {page.number+1} has {len(images)} images and {len(texts)} text blocks")
```

```python
for size in sizes:
    pdf_path = f"data/{year}/{year}_{quarter}_{size}.pdf"
    print(f"{pdf_path}")
    
    doc = pymupdf.open(pdf_path)
    for page in doc:
      images = page.get_images()
      print(f"Page {page.number+1} has {len(images)} images")
      pix = pymupdf.Pixmap(doc, images[0][0])
      pix.save(f"{pic_path}{year}_{quarter}_{size}_page{page.number+1}.png")
      pix = None
```
